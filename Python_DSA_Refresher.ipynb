```python
#Question 4
character=(input("Enter a character: "))
if(character>='A' and character<='Z'):
    print("The character is an uppercase letter.")
elif(character>='a' and character<='z'):
    print("The character is a lowercase letter.")
else:
    print("not a letter.")
```
